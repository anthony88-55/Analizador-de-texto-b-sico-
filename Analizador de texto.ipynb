### Resolución

Ingreso del usuario:

```python
frase = input("Ingrese un texto de minimo 10 palabras")
```

1. Contar el número total de caracteres en el texto

```python
num_caracteres = len(frase)
num_caracteres
```

2. Contar el número de caracteres sin incluir los espacios

```python
num_caracteres_sin_espacio = len(frase.replace(" ", ""))
num_caracteres_sin_espacio
```

3. Contar la cantidad de vocales que hay en el texto

```python
total_a = frase.count("a")
total_e = frase.count("e")
total_i = frase.count("i")
total_o = frase.count("o")
total_u = frase.count("u")

total_vocales = total_a + total_e + total_i + total_o + total_u

total_vocales
```

```python
lista = list(frase)
vocales = ["a", "e", "i", "o", "u"]

conteo_vocales = 0

for caracter in lista:
    
    if caracter in vocales:
        conteo_vocales += 1

conteo_vocales
```

4. Contar el número total de palabras en el texto ingresado

```python
total_palbras = len(frase.split())
total_palbras
```

5. Eliminar la primera palabra

```python
sin_primera_palabra = frase.split()

sin_primera_palabra = " ".join(sin_primera_palabra[1:])
sin_primera_palabra
```

6. Reemplazar todos los espacios por guiones medios (-)

```python
con_guiones = frase.replace(" ", "-")

con_guiones
```

7. Cambia las mayúsculas a minúsculas y las minúsculas a mayúsculas

```python
cambio_Mayus_min = frase.swapcase()
cambio_Mayus_min
```

Muéstrale al usuario los resultados del análisis de su texto.

```python
print(

f"""
Tu frase es: {frase}.

Numero total de caracteres: {num_caracteres}.

Numero de caracteres sin espacios: {num_caracteres_sin_espacio}.

Total de vocales en la frase: {conteo_vocales}.

Numero de palabras: {total_palbras}.

Frase sin la primera palabra: {sin_primera_palabra}.

Frase con guiones: {con_guiones}.

Frase con mayuscula y miniscula: {cambio_Mayus_min}
"""
)
```
